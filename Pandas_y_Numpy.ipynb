```python
import pandas as pd
import numpy as np
import os
```

```python
# ---------------------------
# 1) Cargar y verificar
# ---------------------------
df = pd.read_csv("ecommerce_raw.csv", sep=",")
print("Shape:", df.shape)
print(df.head(5).to_string(index=False))
print(df.columns.tolist())
```

```python
# ---------------------------
# 2) Importante: convertir numéricas
# ---------------------------
# En este dataset real, varias columnas numéricas vienen como object/string
for col in ["edad", "precio_base", "descuento", "rating", "stock", "agrego_carrito"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)
```

```python
# ---------------------------
# 3) Dimensiones con Pandas y NumPy
# ---------------------------
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

arr = df.to_numpy()
print("Array shape:", arr.shape)
```

```python
# ---------------------------
# 4) Tamaño del archivo
# ---------------------------
tamanio_bytes = os.path.getsize("ecommerce_raw.csv")
print("Bytes:", tamanio_bytes)
print("KB:", tamanio_bytes / 1024)
print("MB:", tamanio_bytes / 1024**2)

df.info(memory_usage="deep")
```

```python
# ---------------------------
# 5) Inventario de variables
# ---------------------------
identificadores = ["interaccion_id", "usuario_id", "producto_id"]
numericas = ["edad", "precio_base", "descuento", "precio_pagado", "rating", "stock", "visitas_previas", "compras_previas", "cantidad"]
categoricas = ["provincia", "categoria", "subcategoria", "marca", "dispositivo"]
temporales = ["fecha", "hora"]
binarias = ["agrego_carrito", "compra"]


print("Identificadores:", identificadores)
print("Numéricas:", numericas)
print("Categóricas:", categoricas)
print("Temporales:", temporales)
print("Binarias:", binarias)
```

```python
# ---------------------------
# 6) Tipos de datos
# ---------------------------
print(df.dtypes)

# comentario interpretativo
# edad / precio_base / descuento aparecen como numéricas luego de la conversión.
# Si alguna columna sigue como object, hay que revisarla (puede contener texto o valores raros).

# ---------------------------
# 7) Pandas vs NumPy: selección
# ---------------------------
edad_pandas = df["edad"]
edad_numpy = df["edad"].to_numpy()

print(type(edad_pandas))
print(type(edad_numpy))
print(edad_pandas.head())
print(edad_numpy[:5])

# Pandas conserva nombre, índice y métodos; NumPy devuelve un array puro.
```

```python
# ---------------------------
# 8) Estadística básica
# ---------------------------
print("Promedio edad:", df["edad"].mean())
print("Min edad:", df["edad"].min())
print("Max edad:", df["edad"].max())
print("Std edad:", df["edad"].std())

print("NumPy promedio:", np.mean(df["edad"].dropna()))
print("NumPy min:", np.min(df["edad"].dropna()))
print("NumPy max:", np.max(df["edad"].dropna()))
print("NumPy std:", np.std(df["edad"].dropna()))
```

```python
# ---------------------------
# 9) Estadística completa
# ---------------------------
print(df["edad"].describe())

q1, q2, q3 = np.percentile(df["edad"].dropna(), [25, 50, 75])
print("Q1, mediana, Q3:", q1, q2, q3)

# Los cuartiles dividen los datos en 4 partes iguales.
# Q1 = 25%, Q2 = mediana, Q3 = 75%.
```

```python
# ---------------------------
# 10) Valores faltantes
# ---------------------------
nulos = df.isnull().sum()
print(nulos.sort_values(ascending=False).head(10))
print("Total nulos:", df.isnull().sum().sum())

# 3 variables con más faltantes
print(nulos.sort_values(ascending=False).head(3))
```

```python
# ---------------------------
# 11) Categorías
# ---------------------------
print("Valores únicos de categoria:", df["categoria"].unique()[:10])
print("Cantidad de categorías:", df["categoria"].nunique())
print(df["categoria"].value_counts().head(10))

valores, cantidades = np.unique(df["categoria"].dropna().astype(str), return_counts=True)
print(dict(zip(valores, cantidades)))
```

```python
# ---------------------------
# 12) Dispositivos
# ---------------------------
print("Valores distintos de dispositivo:", df["dispositivo"].unique())
print(df["dispositivo"].value_counts())

# Ejemplo de revisión:
# si aparecen valores extraños, se investigan, pero no se corrigen aún.
```

```python
# ---------------------------
# 13) Rangos sospechosos
# ---------------------------
for col in ["edad", "rating", "descuento", "precio_base"]:
    print(col, "min:", df[col].min(), "max:", df[col].max())

# Por ejemplo: si edad tiene 0 o valores imposibles, eso hay que revisarlo.
# Si precio_base o descuento son negativos o absurdos, también.
```

```python

# ---------------------------
# 14) Duplicados
# ---------------------------
print("Filas duplicadas:", df.duplicated().sum())
print(df[df.duplicated(keep=False)].head(10).to_string(index=False))
```

```python
# ---------------------------
# 15) Variable binaria
# ---------------------------
print("Valores únicos compra:", df["compra"].unique())
print(df["compra"].value_counts())
print("Valores únicos agrego_carrito:", df["agrego_carrito"].unique())
print(df["agrego_carrito"].value_counts())

# Si hay valores distintos de 0 y 1, hay que revisarlos antes de modelar.
```

```python
# ---------------------------
# 16) NumPy y vectorización
# ---------------------------
edades = np.array([20, 25, 30, 35, 40])
print("Edades + 10:", edades + 10)

# Importante porque NumPy hace la operación sobre todo el vector a la vez,
# sin bucles Python, y es mucho más eficiente.
```

```python
# ---------------------------
# 17) Diagnóstico de una variable
# ---------------------------
var = "precio_base"
serie = df[var]

print("Tipo:", serie.dtype)
print("Válidos:", serie.notna().sum())
print("Faltantes:", serie.isna().sum())
print("Media:", serie.mean())
print("Mediana:", serie.median())
print("Mínimo:", serie.min())
print("Máximo:", serie.max())
print("Desviación:", serie.std())

# Qué investigar:
# - valores atípicos
# - negativos
# - nulos
# - si hay datos inconsistentes con el negocio
```

```python
# ---------------------------
# 18) Comparación argumentada
# ---------------------------
tabla = pd.DataFrame({
    "Operación": [
        "Dimensiones",
        "Primeras filas",
        "Tipos",
        "Promedio",
        "Mínimo/Máximo",
        "Desvío estándar",
        "Valores únicos",
        "Frecuecias",
        "Datos tabulares"
    ],
    "Más natural": [
        "Cualquiera",
        "Pandas",
        "Pandas",
        "Cualquiera",
        "Cualquiera",
        "Cualquiera",
        "Pandas",
        "Pandas",
        "Pandas"
    ],
    "Justificación": [
        "df.shape y array.shape son equivalentes",
        "df.head() es más legible que array[:5]",
        "df.dtypes da contexto de columnas",
        "np.mean() es muy bueno en arrays; Series.mean() es claro",
        "np.min()/np.max() y min()/max() son equivalentes",
        "std() y np.std() se usan según el contexto",
        "unique()/nunique() están más orientados a tablas",
        "value_counts() es muy claro para variables categóricas",
        "Pandas está pensado para tablas con nombres de columnas"
    ]
})
print(tabla)
```
